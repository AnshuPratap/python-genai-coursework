### Safe Shopping Cart

Create a program that:
1. Has a cart list --> cart = []
2. Runs a loop asking user to enter prices.
3. Stops when user enter 'q'
4. Inside the loop:
    1. Convert input to a float
    2. Handle ValueError if user enter invalid input
    3. Raise custom exception if price is negative
5. at the end, print:
    1. Total Items
    2. Total bill

```python
cart = []
total = 0
class NegativeValueError(Exception):
    pass
while True:
    price = input("Enter prices (or q to exit): ")
    
    if price.lower() == "q":
        break

    try:
        price = float(price)
        total += price
        if price < 0:
            raise NegativeValueError("Price is Negative! Not allowed!")
        cart.append(price)
    except ValueError:
        print("Enter Valid Input!!")
    except NegativeValueError as e:
        print(e)
print(cart)
print("total items:", len(cart))
print("total_bill:", total)
```
