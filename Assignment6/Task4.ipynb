### File Reader using Exception Handling

1. Ask the user for filename.
2. Try to open and read the file.
3. Handle:
    1. FileNotFounfError
    2. PermissionError
4. If successsful, print first 3 lines of the file.
5. Use finally to print:
"File operation attempted."

```python
#type "data.txt" for existing file
file_name = input("Enter the filename:")
```

```python
try:
    with open(file_name,"r") as f:
        data = f.readlines()[:3]
        for line in data:
            print(line.strip())
except FileNotFoundError:
    print("No such file or directory.")
except PermissionError:
    print("Permission not granted!!")
finally:
    print("File operation attempted.")
```
