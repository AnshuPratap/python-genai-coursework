### Custom Exception : Age Validator

1. Write a function check_age(age) that:
    1. Raises a custom exception ValueError("Age must be between 1 and 120") age is out of range
2. In your main code:
    1. Take age input from the user.
    2. Use try-except to catch and print the custom error message.

```python
def check_age(age):
    if age < 1 or age > 120:
        raise ValueError("Age must be between 1 and 120")
    print("Age is Valid")
```

```python
try:
    age = int(input("Enter your age: "))
    check_age(age)
except ValueError as e:
    print(e)
```
