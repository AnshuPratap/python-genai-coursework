### Bill Calculator with Error Handling

Given a list of product prices:
prices = [120,350,'abc',500,-200,800]

Write a code that:
1. Iterates through the list.
2. Tries to add only valid(positive numerical) prices to the total.
3. Handles:
    1. TypeError if value is not a number
    2. Custom exception using raise ValueError("Negative price not allowed")
4. Prints the running total.

Expected behaviour:
Skip invalid items but continue processing.

```python
prices = [ 120, 350, 'abc', 500, -200, 800]
```

```python
total = 0
for price in prices:
    try:
        if not isinstance(price, (int, float)):
            raise TypeError("Price must be numeric")
        if price < 0:
            raise ValueError("Negative price not allowed")
        total += price
        print(total)
    except TypeError:
        print("Given Value is not a number")
    except ValueError as e:
        print(e)
```
