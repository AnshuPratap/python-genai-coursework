## TASK 1:  Basic Class & Object Creation

Create a class Product with:
1. Attributes: name,price,category
2. Method: get_info() --> print product details
3. Create Two objects and call get_info()

Extra: Add a method apply_discount(percent) that returns the discounted price

```python
class Product():
    def __init__ (self,name,price,category):
        self.name = name
        self.price = price
        self.category = category
    def get_info(self):
        print(f"The product {self.name} which lies on {self.category} category is of price: {self.price}$")
    def apply_discount(self,percent):
        if not 0 <= percent <= 100:
            raise ValueError("Discount percentage must be in between 0 and 100")
        print(f"After discount total price on your product:{self.price * (1- percent / 100)}$")
```

```python
milk = Product("Milk",10,"Dairy Product")
chair = Product("Chair",500,"Furniture")
```

```python
milk.get_info()
chair.get_info()
chair.apply_discount(15)
```

## TASK2: Constructor and Encapsulation

Modify the Product class:
1. Make price a private attribte(__price)
2. Create getter and setter methods:
    1. get_price()
    2. set_price(new_price) --> should update only if new_price > 0
3. Test modifying price using the setter

```python
class Product():
    def __init__ (self,name,price,category):
        self.name = name
        self.__price = price
        self.category = category
    def get_info(self):
        print(f"The product {self.name} which lies on {self.category} category is of price: {self.__price}$")
    #getter method
    def get_price(self):
        return self.__price
    #setter method
    def set_price(self,new_price):
        if new_price < 0:
            raise ValueError("Price cannot be negative")
        else:
            self.__price = new_price
            print(f"The new price is {self.__price}")
```

```python
mobile = Product("Mobile",1000,"Electronics")
```

```python
mobile.get_info()
```

```python
mobile.get_price()
```

```python
mobile.set_price(1200)
```

## TASK3: Inheritance(single_level)

Craete a subclass ElectronicProduct that inherits form product.
1. Additional attrbute: warranty_years
2. Override the get_info() method to include warranty info.
3. Craete an object and demonstarte inheritance + overriding

```python
class ElectronicProduct(Product):
    def __init__(self,name,price,category,warranty_years):
        super().__init__(name,price,category)
        self.warranty_years = warranty_years
    #Overriding
    def get_info(self):
        print(f"The product {self.name} which lies on {self.category} category having warranty of {self.warranty_years} years")
```

## TASK6: Magic Methods & Operator Overloading

Add the following to your Product class:
1. __str__ method
2. Returns a readbale string:
    1. Product(name,price,category)
3. Operator Overloading(__add__)
4. Allow: product1 + product2
5. To return the total combined price.
6. Test this with two product objects.

```python
class Product:
    def __init__(self, name, price, category):
        self.name = name
        self.__price = price
        self.category = category
    def __str__(self):
        return f"{self.name} - {self.category} - {self.__price}$"
```

```python
bed = Product("Bed", 100, "Furniture")
print(bed)
```

```python
class Product:
    def __init__(self, name, price, category):
        self.name = name
        self.__price = price
        self.category = category
    def __add__(self,other):
        return Product(self.name + " & " + other.name,self.__price + other.__price,self.category + " & " + other.category)
    def __str__(self):
        return f"{self.name} - {self.category} - {self.__price}$"
```

```python
pepsi = Product("Pepsi", 2, "Beverage")
sprite = Product("Sprite", 3, "Beverage")
cold_drink = pepsi + sprite
print(cold_drink)
```

Create Two classes:
1. class: Inventory
2. Attributes: 
    1. product(list to store product objects)
3. Methods:
    1. add_product(product)
    2. remove_product(name)
    3. get_total_value() --> sums prices of all products
    4. show_all_products() --> prints info for each product
4. class: Store
5. Attributes:
    1. store_name
    2. inventory --> an inventory object
6. Methods: 
    1. add_new_product() --> takes input & creates Product object.
    2. show_summary() --> prints total items & value
7. Test the system by:
    1. Creating a Store object
    2. Addig 3 Products
    3. Showing summary
    4. Using __add__ to combine prices of two products

```python
class Product:
    def __init__(self,name,price):
        self.name = name
        self.price = price
    def __add__(self,other):
        return Product(self.name + " & " + other.name, self.price + other.price)
```

```python
class Inventory:
    def __init__(self):
        self.products = []

    def add_product(self,product):
        self.products.append(product)

    def remove_product(self,name):
        for product in self.products:
            if product.name == name:
                self.products.remove(product)
                return
        print("Product not in the list!")
        
    def total_price(self):
        total = sum(product.price for product in self.products)
        return total

    def show_all_products(self):
        for product in self.products:
            print(f"Product name : {product.name.capitalize()}, Price : {product.price}$")
```

```python
class Store(Inventory):
    def __init__(self,name):
        super().__init__()
        self.name = name
    def add_new_product(self,product):
        self.products.append(product)

    def show_summary(self):
        print(f"Store name: {self.name}")
        print(f"Total Items: {len(self.products)}")
        print(f"Total value : {self.total_price()}$")
```

```python
store = Store("Avengers Store")
store.add_new_product(Product("Iron Man Suit", 1000000))
store.add_new_product(Product("Captain America Shield", 500000))
store.add_new_product(Product("Thor's Hammer", 750000))
```
